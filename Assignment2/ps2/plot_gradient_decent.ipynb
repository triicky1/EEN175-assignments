```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

```

```python
def function(x):
    f_x = x**2
    df_x = 2*x
    return [f_x, df_x]
```

```python
def function(x):
    f_x = (0.5*np.sin(x) + 3*np.cos(0.34*x))
    df_x = (0.5*np.cos(x) - 3*np.sin(0.34*x)*0.34)
    return [f_x, df_x]
```

```python

x = np.linspace(-20, 20, 200)
fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)

def update(step_size, nr_of_steps, init_x):
    ax.clear()

    ax.plot(x, function(x)[0], linewidth=3)
    x_hat = init_x
    ax.plot(x_hat, function(x_hat)[0], '*', markersize=15)

    for i in range(nr_of_steps):
        f, f_grad = function(x_hat)
        x_hat_new = x_hat - step_size * f_grad
        f_new, f_grad = function(x_hat_new)
        ax.plot([x_hat, x_hat_new], [f, f_new], linewidth=2)  
        ax.plot(x_hat_new, f_new, '*', markersize=15) 
        x_hat = x_hat_new 

    fig.canvas.draw_idle()

interact(update, step_size = widgets.FloatSlider(value=0.1,
                                               min=0,
                                               max=10.0,
                                               step=0.05), 
         nr_of_steps = widgets.IntSlider(value=1, 
                                     step=1, 
                                     min=0, 
                                     max=100),
        init_x = widgets.FloatSlider(value=3,
                                    min=-30,
                                    max=30.0,
                                    step=0.2));
```
