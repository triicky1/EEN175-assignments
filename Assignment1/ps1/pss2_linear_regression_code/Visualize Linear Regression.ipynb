```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, ToggleButtons

# Function to generate data
def generate_data(model_type, coef, intercept, noise_level, n_points=50):
    x = np.linspace(-5, 5, n_points)
    if model_type == 'Linear':
        y_true = coef * x + intercept
    elif model_type == 'Sinusoid':
        y_true = coef * np.sin(x) + intercept
    
    noise = np.random.normal(0, noise_level, size=n_points)
    y_noisy = y_true + noise
    return x, y_noisy, y_true

# Function to fit model
def fit_model(model_type, x, y_noisy):
    if model_type == 'Linear':
        # Linear regression fit
        A = np.vstack([x, np.ones_like(x)]).T
        coef_est, intercept_est = np.linalg.lstsq(A, y_noisy, rcond=None)[0]
        y_est = coef_est * x + intercept_est
        return y_est, coef_est, intercept_est
    elif model_type == 'Sinusoid':
        # Fit sinusoid: y ≈ a*sin(x) + b
        A = np.vstack([np.sin(x), np.ones_like(x)]).T
        coef_est, intercept_est = np.linalg.lstsq(A, y_noisy, rcond=None)[0]
        y_est = coef_est * np.sin(x) + intercept_est
        return y_est, coef_est, intercept_est

# Plotting function
def plot_model(model_type, coef, intercept, noise_level):
    x, y_noisy, y_true = generate_data(model_type, coef, intercept, noise_level)
    y_est, coef_est, intercept_est = fit_model(model_type, x, y_noisy)
    
    plt.figure(figsize=(8,5))
    plt.scatter(x, y_noisy, label="Noisy Data", alpha=0.7)
    plt.plot(x, y_true, color="red", label=f"True {model_type} Model")
    plt.plot(x, y_est, 'b--', label=f"Estimated Fit")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(f"{model_type} Regression Visualization\nEst: coef={coef_est:.2f}, intercept={intercept_est:.2f}")
    plt.legend()
    plt.grid(True)
    plt.xlim(-5, 5)
    plt.ylim(-10, 10)
    plt.show()

# Interactive widgets
interact(
    plot_model,
    model_type=ToggleButtons(options=['Linear', 'Sinusoid'], description="Model:"),
    coef=FloatSlider(value=1.0, min=-5, max=5, step=0.1, description="Coef"),
    intercept=FloatSlider(value=0.0, min=-5, max=5, step=0.1, description="Intercept"),
    noise_level=FloatSlider(value=1.0, min=0.0, max=10.0, step=0.1, description="Noise")
)
```
